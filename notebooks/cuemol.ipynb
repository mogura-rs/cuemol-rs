# CueMol in Colab

Install the current public release and read the same agent skill provided by `cuemol skill` and [GitHub](https://github.com/mogura-rs/cuemol-rs/blob/main/skills/cuemol/SKILL.md). A CPU runtime is sufficient for these cells. The installer verifies the release checksum and writes its files beneath `.cuemol`.


```python
%%bash
set -euo pipefail
curl -LsSf https://raw.githubusercontent.com/mogura-rs/cuemol-rs/main/install.sh | CUEMOL_INSTALL_DIR="$PWD/.cuemol/bin" sh
./.cuemol/bin/cuemol --version
```

```python
from pathlib import Path
import subprocess
from IPython.display import Markdown, display

cuemol = str(Path(".cuemol/bin/cuemol").resolve())
skill = subprocess.check_output([cuemol, "skill"], text=True)
Path(".cuemol/SKILL.md").write_text(skill, encoding="utf-8")
display(Markdown(skill))
```

```python
import json

print(subprocess.check_output([cuemol, "-c", "print(21 * 2);"], text=True), end="")
tools = json.loads(subprocess.check_output([cuemol, "mcp", "tools"], text=True))["tools"]
print(f"Available MCP tools: {len(tools)}")
print(", ".join(tool["name"] for tool in tools[:12]))
```

## Continue with your structures

Use `cuemol run commands.js`, `cuemol render structure.pdb image.png`, or a persistent `cuemol mcp serve` process for headless work. Separate CLI calls do not share a scene. The native GUI needs a desktop and does not open inside a Colab notebook. See the [user guide](https://github.com/mogura-rs/cuemol-rs/blob/main/docs/usage.md) for commands and rendering options.

A source installation is also available with `pip install git+https://github.com/mogura-rs/cuemol-rs.git` after installing Rust, a C++20 compiler, and the documented native build dependencies. The release installer used above avoids that compilation step.
